## Global versus local interpretability
- **Global:** aims to provide an understanding of **the overall behavior** of the trained model across the entire dataset.
  - For instance, questions like "Which features are generally important for the model's predictions?" or "What kind of interactions between features occur across the dataset?"
  - It can be challenging, especially for complex models with numerous parameters. Humans often struggle to grasp the intricate workings of such models holistically.
- **Local**: focuses on explaining **individual predictions**. It zooms in on a single prediction (or a small group of them) and analyzes the model's behavior in the neighborhood of that instance.
  - For example, if a model predicts a high risk of cancer for a patient, local interpretability would help understand why that specific prediction was made, considering the individual's features. This approach might reveal that specific combinations of features led to the prediction, even if the model's overall behavior is complex and difficult to comprehend globally.

# Dataset: risk factors for cervical cancer

The cervical cancer dataset contains indicators and risk factors for predicting whether a woman will get cervical cancer. The features include demographic data (such as age), lifestyle, and medical history. The data can be downloaded from the UCI Machine Learning repository and is described by Fernandes, Cardoso, and Fernandes (2017).

The subset of data features used here are:

- Age in years
- Number of sexual partners
- First sexual intercourse (age in years)
- Number of pregnancies
- Smoking yes or no
- Smoking (in years)
- Hormonal contraceptives yes or no
- Hormonal contraceptives (in years)
- Intrauterine device yes or no (IUD)
- Number of years with an intrauterine device (IUD)
- Has patient ever had a sexually transmitted disease (STD) yes or no
- Number of STD diagnoses
- Time since first STD diagnosis
- Time since last STD diagnosis
- The biopsy results “Healthy” or “Cancer”. Target outcome.

```python
import pandas as pd
from sklearn.impute import SimpleImputer

cervical = pd.read_csv('https://cs.calvin.edu/courses/data/202/fsantos/datasets/cervical.csv', na_values='?', keep_default_na=False)

# Drop specified columns
cervical = cervical.drop(columns=['Citology', 'Schiller', 'Hinselmann'], errors='ignore')

# Select specific columns
cervical = cervical[['Age', 'Number.of.sexual.partners', 'First.sexual.intercourse',
                    'Num.of.pregnancies', 'Smokes', 'Smokes..years.', 'Hormonal.Contraceptives',
                    'Hormonal.Contraceptives..years.', 'IUD', 'IUD..years.', 'STDs', 'STDs..number.',
                    'STDs..Number.of.diagnosis', 'STDs..Time.since.first.diagnosis', 'STDs..Time.since.last.diagnosis', 'Biopsy']]

# Replace 'Biopsy' values
cervical['Biopsy'] = cervical['Biopsy'].replace({'Healthy': 0, 'Cancer': 1})

# NA imputation using mode
imputer = SimpleImputer(missing_values=pd.NA, strategy='most_frequent')
cervical_numeric = cervical.select_dtypes(include=['number'])
cervical[cervical_numeric.columns] = imputer.fit_transform(cervical_numeric)

cervical
```

```python
import pandas as pd
from sklearn.impute import SimpleImputer
from imblearn.over_sampling import SMOTE

# ... (your existing code) ...

# Separate features (X) and target (y)
X = cervical.drop('Biopsy', axis=1)
y = cervical['Biopsy']

# Apply SMOTE for class balancing
smote = SMOTE(random_state=42)  # You can specify a different random state
X_resampled, y_resampled = smote.fit_resample(X, y)
```

```python
from sklearn.model_selection import train_test_split, cross_validate, StratifiedKFold
from sklearn.metrics import accuracy_score, classification_report
from imblearn.ensemble import BalancedRandomForestClassifier
from sklearn.metrics import precision_score, recall_score, make_scorer

# Use the Balanced version of Random Forest (in library imblearn)
brf = BalancedRandomForestClassifier(random_state=42)
brf.fit(X_resampled, y_resampled)

# Use StratifiedKFold for cross-validation
cv = StratifiedKFold(n_splits=20, shuffle=True, random_state=42)

# Create scorers for precision and recall
scoring = {
    'precision_class_0': make_scorer(precision_score, pos_label=0),
    'precision_class_1': make_scorer(precision_score, pos_label=1),
    'recall_class_0': make_scorer(recall_score, pos_label=0),
    'recall_class_1': make_scorer(recall_score, pos_label=1)
}

# Perform cross-validation
cv_results = cross_validate(brf, X_resampled, y_resampled, scoring=scoring, cv=cv, n_jobs=-1)

# Extract and print results
for metric in scoring.keys():
    print(f"{metric} - Mean: {cv_results['test_' + metric].mean()})")
```

Let's focus on this person:

```python
person = cervical.iloc[[22]]
prediction = brf.predict(person.drop('Biopsy', axis=1))
print(person)
print(prediction)
```

# Individual Conditional Expectation (ICE)

Remember the Partial Dependence Plot? We were averaging the prediction over a fixed value of a variable.

What if we just use one data row, though? This is the Individual Conditional Expectation (ICE).

```python
import matplotlib.pyplot as plt
import numpy as np

# Function to calculate ICE for a specific row
def plot_ice(model, X, feature_name, row_index, feature_range=None):
    """
    Plot ICE for a specific feature and row.

    Parameters:
    - model: Trained machine learning model.
    - X: DataFrame containing features.
    - feature_name: Name of the feature for ICE.
    - row_index: Index of the row for which ICE is plotted.
    - feature_range: Range of values to vary the feature (optional).
    """
    # Select the specific row
    row = X.iloc[row_index].copy()

    # Generate a range for the selected feature
    if feature_range is None:
        feature_range = np.linspace(X[feature_name].min(), X[feature_name].max(), 50)

    # Store predictions
    predictions = []

    for value in feature_range:
        # Modify the row's feature value
        row[feature_name] = value
        row_df = pd.DataFrame([row], columns=X.columns)

        # Predict using the model
        pred = model.predict_proba(row_df)[0][1]  # Assuming binary classification
        predictions.append(pred)

    # Plot ICE
    plt.figure(figsize=(8, 5))
    plt.plot(feature_range, predictions, label=f'Row {row_index}')
    plt.xlabel(feature_name)
    plt.ylabel('Prediction (Probability of Class 1)')
    plt.title(f'ICE Plot for Feature: {feature_name} (Row {row_index})')
    plt.legend()
    plt.grid(True)
    plt.show()

# Example: Plot ICE for Feature 1 in Row 10
plot_ice(brf, X, feature_name="Age", row_index=22)
```

# Local Surrogate Model (LIME)

"Local Interpretable Model-agnostic Explanations (LIME)". The idea is of a surrogate interpretable model...

Answers the question: "among similar instances of this data, what specifically makes a difference in prediction?"
- or: "among persons similar to those, what makes a difference from the absence to presence of cancer?"

image.png

Steps:
- **1. Select an Instance to Explain**
- **2. Generate Perturbed Samples**
  - Create a dataset of perturbed versions of the selected instance by slightly modifying its feature values.
  - These perturbations can be generated by:
    - Sampling around the instance using random noise (for numerical features).
    - Sampling from the distribution of categorical features in the dataset.
- **3. Predict Using the Black-box Model**
  - Pass each of the perturbed samples through the black-box model and record its predictions (e.g., class probabilities for classification models).
- **4. Assign Weights to Perturbed Samples**
  - Assign a weight to each perturbed sample based on its similarity to the original instance.
  - This is done using a kernel function (e.g., Gaussian kernel) that assigns higher weights to samples closer to the original instance in feature space.
- **5. Fit an Interpretable Model Locally**
  - Fit a simple interpretable model (e.g., linear regression, decision tree) using the perturbed samples, their weights, and the black-box model's predictions as the target.
  - The interpretable model captures the local decision boundary of the black-box model around the instance.
- **6. Extract Feature Importance**
  - The coefficients of the interpretable model represent the importance of each feature for the prediction of the selected instance.
- **7. Present the Explanation**
  - The final explanation is presented as a set of feature importance scores, indicating how each feature contributed to the model's prediction for the instance.

```python
!pip install lime
```

```python
import lime
import lime.lime_tabular
import numpy as np

# Convert X to a NumPy array with only numerical features
X_numpy = X.select_dtypes(include=np.number).to_numpy()

# Initialize the LIME explainer using the NumPy array
explainer = lime.lime_tabular.LimeTabularExplainer(
    training_data=X_numpy,         # Use the NumPy array
    feature_names=X.select_dtypes(include=np.number).columns.tolist(),  # Feature names from numerical columns
    class_names=['Healthy', 'Cancer'],
    mode='classification',
    random_state=42
)

# Ensure instance is also a NumPy array and has the correct shape
instance = X.select_dtypes(include=np.number).iloc[[22]].to_numpy()

# Explain the prediction
explanation = explainer.explain_instance(
    data_row=instance[0],
    predict_fn=brf.predict_proba
)

# Display the explanation
explanation.show_in_notebook(show_table=True)
```

# SHAP (SHapley Additive exPlanations)

> The Shapley value, coined by Lloyd Shapley (1953), is a method for assigning payouts to players depending on their contribution to the total payout. Players cooperate in a coalition and receive a certain profit from this cooperation.

Based on [this example](https://towardsdatascience.com/shap-shapley-additive-explanations-5a2a271ed9c3): Let's say we have friends A, B, C and D and we want to estimate how much A contributed. We can do this:

image.png

However, this only calculates the contribution of A against B, C, and D. We can, however, calculate the contribution of A over all possible configurations - and get the mean:

image.png

SHAP (SHapley Additive exPlanations) was proposed and implemented by Lundberg and Lee (2017) and is a way to efficiently calculate Shapley values for predictions.

```python
import shap

# Explain the model's predictions using SHAP
explainer = shap.Explainer(brf)
shap_values = explainer(person)

# Visualize SHAP values
shap.waterfall_plot(shap_values[0,:,1])
```

# Counterfactual Explanations

Counterfactual explanations provide a way to understand model predictions by answering the question: "What minimal change in the input features would lead to a different prediction?"

- For example, for a prediction that someone is classified as having cancer (Biopsy=1), a counterfactual explanation might state: "If the number of sexual partners were reduced by 1, the model would classify the person as healthy (Biopsy=0)."

We will use the library `dice-ml` for that.

```python
pip install dice-ml
```

```python
import dice_ml
from dice_ml.utils import helpers

cervical_balanced = pd.concat([X_resampled, y_resampled], axis=1)

# Initialize DiCE data interface
data_interface = dice_ml.Data(dataframe=cervical_balanced,
                              continuous_features=['Age', 'Number.of.sexual.partners',
                                                  'First.sexual.intercourse', 'Num.of.pregnancies'],
                              outcome_name='Biopsy')

# Initialize DiCE model interface
model_interface = dice_ml.Model(model=brf, backend='sklearn', model_type='classifier')

# Create DiCE explainer
exp = dice_ml.Dice(data_interface, model_interface)

# Generate counterfactual explanations
# Create a new DataFrame without the 'Biopsy' column for query instance
query_instance = person.drop('Biopsy', axis=1)
dice_exp = exp.generate_counterfactuals(query_instance,
                                       total_CFs=5,
                                       desired_class="opposite")

# Visualize the counterfactual explanations
dice_exp.visualize_as_dataframe()
```
