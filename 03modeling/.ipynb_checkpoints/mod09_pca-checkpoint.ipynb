# Unsupervised Learning

- Algorithms analyze and structure data **without predefined labels or outcomes**. The primary goal is to **discover hidden patterns, groupings, or relationships** within the data without any prior guidance or correct answers.

Among the main types of tasks, we find:
1. **Dimensionality Reduction**: Reducing the number of features while preserving important information. This is used for data visualization, noise reduction, and preparing data for supervised learning.
  - Examples: Principal Component Analysis (PCA), Singular Value Decomposition (SVD), t-SNE
2. **Clustering**: Grouping similar data points together. Widely used for customer segmentation, image segmentation, and document grouping.
  - We'll see some of them on next class.
3. **Probability Estimation**: Estimating the likelihood of different outcomes or the probability distribution of data.
  - Examples: Gaussian Mixture, Kernel Density Estimation, Bayesian Networks
4. **Anomaly Detection**: Identifying unusual patterns or outliers that don't conform to expected behavior. Applications include fraud detection, network security, and medical diagnostics.
  - Examples: Isolation Forest and One-Class SVM.
5. **Association Learning**: Discovering relationships or associations between variables, often used in market basket analysis to find frequently bought together items.
  - Examples: Apriori and Eclat.

# PCA - Principal Component Analysis

PS: not [this PCA](https://pcanet.org/)

**There are so many uses! Here is a summary given to us by AI:**

### 1. **Dimensionality Reduction for High-Dimensional Data**

   **Example**: A dataset with thousands of features, such as gene expression data in bioinformatics or image data for computer vision, can be difficult to handle in machine learning models. High-dimensional data increases the risk of overfitting and may slow down algorithms like logistic regression or k-nearest neighbors (k-NN).
   
   - **PCA Solution**: By applying PCA, you can reduce the number of features while retaining most of the important variance in the data. This reduction allows models to train more efficiently and generalize better by removing redundant or noisy features.
   - **Use Case**: In genomics, PCA is often applied to reduce tens of thousands of gene expression levels into a few hundred principal components that explain most of the variance. This compressed data is then used in clustering (e.g., k-means) or classification tasks (e.g., cancer detection).

### 2. **Image Compression and Feature Extraction**

   **Example**: In image recognition tasks, images typically contain thousands or millions of pixels, resulting in extremely high-dimensional data. However, not all pixels contribute equally to distinguishing objects in images.
   
   - **PCA Solution**: PCA is applied to reduce the dimensionality of image data while retaining important features. The top principal components represent the main patterns or edges in the images, making it easier for machine learning models to classify objects.
   - **Use Case**: For facial recognition, PCA is used to create **eigenfaces**, which are the principal components representing key variations in facial images. These eigenfaces can be used to train models to recognize different faces more efficiently than using the raw pixel data.

### 3. **Removing Multicollinearity in Regression Models**

   **Example**: In linear regression models, if the input features are highly correlated (multicollinear), it can lead to unstable models, where small changes in the data can result in large changes in the predictions.
   
   - **PCA Solution**: PCA can be used to transform the correlated variables into a set of uncorrelated principal components. By using these components as features, the issue of multicollinearity is eliminated, leading to more stable and interpretable regression models.
   - **Use Case**: In economics, when trying to predict housing prices using variables like square footage, number of rooms, and lot size, these features may be correlated. Applying PCA transforms them into uncorrelated components, which can be used in a linear regression model to predict housing prices more reliably.

### 4. **Noise Reduction for Time-Series Data**

   **Example**: In time-series data, especially in sensor data (e.g., IoT devices), there is often a lot of noise that can reduce the accuracy of machine learning models.
   
   - **PCA Solution**: PCA can be used to reduce the impact of noise by identifying and removing components that capture very little variance, which are typically associated with noise. The remaining components capture the meaningful patterns in the data.
   - **Use Case**: In financial markets, PCA is applied to reduce noise in stock prices. By focusing on the principal components that capture real trends in stock movement, analysts can create better predictive models for stock price forecasting.

### 5. **Reducing Feature Space for Clustering Algorithms**

   **Example**: Clustering algorithms like k-means or hierarchical clustering can struggle with high-dimensional data due to the curse of dimensionality, where distance metrics become less meaningful as the number of features increases.
   
   - **PCA Solution**: By applying PCA to reduce the dimensionality of the data, clustering algorithms can perform better, as the data is transformed into a space where the main variations are more pronounced.
   - **Use Case**: In customer segmentation, businesses may have datasets with many features such as purchase history, demographic data, and behavioral data. By applying PCA before clustering, the reduced feature space makes it easier to identify distinct customer segments.

### 6. **Preprocessing Text Data for Topic Modeling**

   **Example**: In natural language processing (NLP), text data is often represented as high-dimensional feature vectors, such as term frequency-inverse document frequency (TF-IDF) matrices. These vectors can have thousands of dimensions, making it hard to apply machine learning algorithms directly.
   
   - **PCA Solution**: PCA can reduce the dimensionality of these vectors by capturing the main themes (principal components) in the text. This makes the data easier to handle and can improve the performance of topic modeling algorithms like Latent Dirichlet Allocation (LDA) or clustering.
   - **Use Case**: For news article classification, a dataset with thousands of unique words is transformed into principal components that represent the core topics in the articles. This reduced data is then fed into a classification model to predict the category of each article (e.g., sports, politics, etc.).

### 7. **Preprocessing for Anomaly Detection**

   **Example**: In anomaly detection tasks, such as identifying fraudulent transactions, the dataset may contain many irrelevant or redundant features that obscure the signal from anomalies.
   
   - **PCA Solution**: PCA helps reduce the dimensionality of the dataset, focusing on components that capture the main variance while ignoring noise. This preprocessing step allows anomaly detection algorithms to focus on the most relevant features, improving detection accuracy.
   - **Use Case**: In credit card fraud detection, PCA is applied to transaction data to reduce the dimensionality. After PCA, clustering algorithms or one-class SVM can be used more effectively to identify anomalous (potentially fraudulent) transactions.

### 8. **Reducing Overfitting in Complex Models**

   **Example**: In models like neural networks or decision trees, having too many features can increase the risk of overfitting, where the model learns noise rather than general patterns.
   
   - **PCA Solution**: PCA reduces the number of input features, preventing the model from overfitting to irrelevant details in the data. The remaining principal components represent the most informative aspects of the data.
   - **Use Case**: In image classification tasks, neural networks trained on high-dimensional pixel data may overfit due to the large number of features. Applying PCA before feeding the images to the neural network reduces the dimensionality and helps the model generalize better to unseen data.

### 9. **Speeding Up Training of Machine Learning Models**

   **Example**: In some machine learning models, especially those involving kernel methods (e.g., support vector machines) or ensemble methods (e.g., random forests), the training time increases significantly with the number of features.
   
   - **PCA Solution**: By reducing the dimensionality of the dataset, PCA can make training faster without losing significant information. This is especially useful when working with large datasets with many features.
   - **Use Case**: In document classification tasks where each document is represented by thousands of word features, applying PCA before training an SVM can greatly reduce the computational burden while maintaining high classification accuracy.

## How does it work?

Explaining it for an expert mathematician: **"PCA is the calculation of the eigenvectors and eigenvalues of the covariance matrix of the features"**

![](https://cs.calvin.edu/courses/data/202/fsantos/img/blinking-what.gif)

Let's go slowly.

- We need to "rotate" our data points to a view that will show the most detail we want.

![](https://cs.calvin.edu/courses/data/202/fsantos/img/rat-spinning.gif)

- **Eigenvectors** are the vectors that specify this rotation.


````python
import plotly.express as px

# Create a 3D scatter plot using Plotly
fig = px.scatter_3d(
    df,
    x='sepal length (cm)',
    y='petal length (cm)',
    z='sepal width (cm)',
    color='target',  # Color the points based on the target variable (species)
    title='Iris Dataset 3D Scatter Plot',
    width=1000,  # Set the width of the figure in pixels
    height=600  # Set the height of the figure in pixels
)
# Update marker size
fig.update_traces(marker=dict(size=3))  # Adjust size as needed, e.g., 3 for smaller markers

fig.show()
````

- The **eigenvectors of the covariance matrix** are the best ones we have!
  - They are the ones in which we have the most variance of the data - i.e., "most detail"
  - These are, exactly, the **PRINCIPAL COMPONENTS**.
  - But, remember - in our example here, we're talking about "5D" data, so it is like a "5D rotation" (whatever that means)

````python
import numpy as np
from sklearn.preprocessing import StandardScaler

# Standardize the data (important for PCA)
x = StandardScaler().fit_transform(df)

# Calculate the covariance matrix
covariance_matrix = np.cov(x.T)

# Calculate the eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(covariance_matrix)

# Print the eigenvectors
print("Eigenvectors of the covariance matrix:\n", eigenvectors)
````

- The **eigenVALUES** corresponding to these eigenvectors indicate **how much variation** there is in each direction rotated.

````python
# Print the eigenvalues
print("Eigenvalues of the covariance matrix:\n", eigenvalues)
````

- So, now, we can **choose the eigenvalues that have the biggest variation**, and their corresponding eigenvectors, to rotate our data in the best direction.
  - In our example above, the first and second ones!

````python
# Select the top 2 eigenvectors corresponding to the largest eigenvalues
sorted_indices = np.argsort(eigenvalues)[::-1]
top_eigenvectors = eigenvectors[:, sorted_indices[:2]]

# Project the data onto the top 2 principal components
projected_data = x.dot(top_eigenvectors)

# Create a 2D scatter plot of the projected data
fig = px.scatter(
    x=projected_data[:, 0],
    y=projected_data[:, 1],
    color=df['target'],
    title='Iris Dataset Projected onto Top 2 Principal Components',
    width=800,
    height=600
)
fig.show()
````

- In sum, this is what PCA is doing, when we call it directly:

````python
from sklearn.decomposition import PCA

# Standardize the features
x = StandardScaler().fit_transform(df.drop('target', axis=1))

# Perform PCA with 2 components
pca = PCA(n_components=2)
principalComponents = pca.fit_transform(x)
principalDf = pd.DataFrame(data=principalComponents, columns=['principal component 1', 'principal component 2'])

# Combine PCA results with target variable
finalDf = pd.concat([principalDf, df[['target']]], axis=1)

# Create a 2D scatter plot of the projected data
fig = px.scatter(
    finalDf,
    x='principal component 1',
    y='principal component 2',
    color='target',
    title='Iris Dataset Projected onto Top 2 Principal Components',
    width=800,
    height=600
)
fig.show()
````

# Example 1: image classification

````python
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits

# Load the digits dataset
digits = load_digits()

# Display some sample images
fig, axes = plt.subplots(nrows=2, ncols=5, figsize=(10, 5))
for i, ax in enumerate(axes.flat):
    ax.imshow(digits.images[i], cmap='gray')
    ax.set_title(f"Digit: {digits.target[i]}")
    ax.axis('off')

plt.show()
````

Each image has 8x8 = 64 pixels.



````python
digits.data.shape
````

Should we use those 64 features? Well, we can reduce to 30, for example:

````python
from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA

X = digits.data  # 8x8 images are flattened into a 64-dimensional vector
y = digits.target  # labels for each image (0 to 9)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# We can try to reduce to 30 principal components for simplicity
pca = PCA(n_components=30)
X_train_pca = pca.fit_transform(X_train)
X_test_pca = pca.transform(X_test)

print(X_train_pca.shape)
print(X_test_pca.shape)
````

Now, we can train a Logistic Regressor to classify our images as numbers:

````python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Train a Logistic Regression model on the reduced data
clf = LogisticRegression(max_iter=1000, random_state=42)
clf.fit(X_train_pca, y_train)

# Make predictions and evaluate the model
y_pred = clf.predict(X_test_pca)
accuracy = accuracy_score(y_test, y_pred)

# Display results
print(f"Accuracy with PCA and Logistic Regression: {accuracy:.2f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
````

## **However...**

- While **PCA** has been a popular tool in image processing and was once commonly used in **image classification** tasks to reduce dimensionality and extract features, it is **no longer the primary choice** for most modern image classification tasks. Here's why and what’s commonly used instead.

### 1. Limitations of PCA for Image Classification
PCA has certain limitations that make it less suitable for complex image classification tasks, especially with the large, high-dimensional datasets commonly used today:

- **Linear Transformation**: PCA is a linear method, meaning it only captures linear relationships in the data. Images often have complex, non-linear patterns and textures that can’t be captured effectively by PCA alone.
- **Loss of Spatial Information**: PCA treats each pixel as an independent feature, which ignores the spatial relationships between pixels. This makes it harder to capture structures and patterns, such as edges, textures, and shapes, which are crucial for distinguishing between image classes.
- **Scalability**: PCA can be computationally expensive for high-dimensional images, such as large RGB images, because it requires calculating and decomposing a large covariance matrix.

### 2. What Has Replaced PCA for Image Classification?

In recent years, **Deep Learning**, particularly **Convolutional Neural Networks (CNNs)**, has become the primary approach for image classification. Here’s why CNNs are preferred over PCA:

- **Non-Linear Feature Extraction**: CNNs can capture complex, non-linear relationships in the data by learning multiple layers of features, from simple edges in early layers to complex shapes in deeper layers.
- **Spatial Awareness**: CNNs take into account the spatial structure of images. By using convolutional filters, they capture local patterns, textures, and shapes that are essential for identifying objects in images.
- **Hierarchical Feature Learning**: CNNs learn features at multiple levels of abstraction, making them robust for classification tasks across a wide range of image complexities.

### 3. When is PCA Still Useful in Image Processing?

PCA still has certain roles in image processing and can complement deep learning methods in specific cases:

- **Data Preprocessing**: PCA is sometimes used to reduce dimensionality before training a model, especially to **reduce noise** in the data or to make computation more efficient on high-dimensional data.
- **Feature Reduction for Lightweight Models**: For lightweight models where computational resources are limited (e.g., mobile applications), PCA can be used to reduce the dimensionality of image features extracted from CNNs. This can help in creating faster, less memory-intensive models.
- **Exploratory Data Analysis**: PCA is valuable for visualizing high-dimensional image datasets. By reducing images to 2D or 3D representations, it allows for quick visualization of clusters or patterns.
- **Eigenfaces for Face Recognition**: PCA was famously used in the **Eigenfaces** technique for face recognition, where it helped identify prominent features for distinguishing faces. While deep learning has largely replaced PCA for face recognition, Eigenfaces is still studied as a foundational approach in computer vision.

# Example 2: Text classification

For dealing with text, we need to convert it to a representation called TF-IDF (Term Frequency-Inverse Document Frequency).



Let's load a sample dataset called 20 Newsgroups.

````python
from sklearn.datasets import fetch_20newsgroups
# Load the 20 Newsgroups dataset
categories = ['rec.sport.baseball', 'rec.sport.hockey', 'sci.space', 'comp.graphics']  # Subset of categories for simplicity
newsgroups = fetch_20newsgroups(subset='all', categories=categories, remove=('footers', 'quotes'))

# Extract the titles (subject lines) from each document
titles = []
for doc in newsgroups.data:
    # Find the subject line
    for line in doc.splitlines():
        if line.lower().startswith("subject:"):
            # Remove "Subject:" prefix and strip whitespace
            titles.append(line[len("subject:"):].strip())
            break
    else:
        # If no subject line found, append an empty string or placeholder
        titles.append("(No subject)")
````

````python
print(titles[1])
print(newsgroups.data[1])
````

We will then use `TfidfVectorizer` to get the TF-IDF representation:

````python
from sklearn.feature_extraction.text import TfidfVectorizer

X_text = newsgroups.data
y = newsgroups.target

# Convert text to TF-IDF features
tfidf_vectorizer = TfidfVectorizer(max_features=2000, stop_words='english')
X_tfidf = tfidf_vectorizer.fit_transform(X_text)

# Convert the matrix to a dataframe for better visualization
tfidf_df = pd.DataFrame(X_tfidf.toarray(), columns=tfidf_vectorizer.get_feature_names_out())
tfidf_df.head()
````

The resulting DataFrame contains a row for each document and a column for each unique word in the corpus. The values represent the "TF-IDF score" of each word in each document (the higher it is, the more frequent in the document).

Now, we have 2000 features! We need to reduce that - probably to 100 features.

We are using a variant of PCA called `TruncatedSVD`, which is better for sparse datasets (i.e., lots of zeros/empty values).

````python
from sklearn.decomposition import TruncatedSVD

# Apply TruncatedSVD (PCA for sparse matrices) for dimensionality reduction
svd = TruncatedSVD(n_components=100, random_state=42)
X_svd = svd.fit_transform(X_tfidf)
````

Let's now use a Logistic Regression model:

````python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Split the reduced data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_svd, y, test_size=0.3, random_state=42)

# Train a Logistic Regression model on the reduced data
clf = LogisticRegression(max_iter=1000, random_state=42)
clf.fit(X_train, y_train)

# Make predictions and evaluate the model
y_pred = clf.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

# Display results
print(f"Accuracy with TruncatedSVD and Logistic Regression: {accuracy:.2f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
````

Let's check some examples.

````python
for i in range(5):
  print(f"Original text: {newsgroups.data[i]}")
  print(f"Label: {newsgroups.target_names[y_test[i]]}")
  print(f"Predicted label: {newsgroups.target_names[y_pred[i]]}")
  print("-" * 50)
````

Another cool thing we can do is plot the document headlines by distance:

````python
# Apply TruncatedSVD to reduce to 2 dimensions
svd = TruncatedSVD(n_components=2, random_state=42)
X_svd = svd.fit_transform(X_tfidf)

# Create a DataFrame for Plotly
target_names = [newsgroups.target_names[t] for t in newsgroups.target]
df = pd.DataFrame({
    'Component 1': X_svd[:, 0],
    'Component 2': X_svd[:, 1],
    'Target': target_names[:len(X_svd)],
    'Title': titles[:len(X_svd)]  # In case of mismatch, slice to fit
})

# Create a 2D scatter plot using Plotly Express
fig = px.scatter(df, x='Component 1', y='Component 2', color='Target',
                 hover_data={'Title': True},
                 title="2D Representation of Newsgroup Headlines")

# Show the plot
fig.show()
````

# Example 3: removing noise from neural signal

PS: That roughly is what I (Prof. Santos) did and published [here]().

````python
# Re-import libraries after code execution state reset
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# Set up the synthetic neuron membrane potential data with chaotic low-frequency noise
np.random.seed(42)
time = np.linspace(0, 1, 1000)  # 1 second at 1000 Hz sampling rate
membrane_potential = -70 + 10 * np.sin(2 * np.pi * 50 * time)  # Neuron potential signal
baseline_shifts = np.piecewise(time, [time < 0.3, (time >= 0.3) & (time < 0.6), time >= 0.6], [-2, 5, -3])
chaotic_noise = baseline_shifts + 1.5 * np.sin(2 * np.pi * 0.5 * time)  # Chaotic low-frequency noise
chaotic_signal_with_noise = membrane_potential + chaotic_noise

# Function to create a lagged version of the signal matrix
def create_signal_matrix(signal, lag):
    rows = len(signal) - lag + 1
    return np.array([signal[i:i + lag] for i in range(rows)])

# Function to apply PCA-based denoising on the signal matrix
def pca_denoise(signal, lag, n_components=1):
    # Create a lagged version of the signal matrix
    M = create_signal_matrix(signal, lag)

    # Apply PCA
    pca = PCA(n_components=n_components)
    M_reduced = pca.fit_transform(M)
    M_reconstructed = pca.inverse_transform(M_reduced)

    # Convert back to 1D signal by averaging the diagonals
    denoised_signal = np.mean(M_reconstructed, axis=1)
    return denoised_signal

# Parameters
lag = 20  # Choose an appropriate lag
n_components = 1  # Number of principal components to retain

# Apply PCA-based denoising to the chaotic signal
denoised_pca_signal = pca_denoise(chaotic_signal_with_noise, lag, n_components)

# Plotting the original, noisy, and PCA-denoised signals
plt.figure(figsize=(15, 6))
plt.plot(time, chaotic_signal_with_noise, label="Recorded Signal", color='orange')
plt.plot(time, membrane_potential, label="True Membrane Potential", color='green', linestyle='dashed')
plt.xlabel("Time (s)")
plt.ylabel("Membrane Potential (mV)")
plt.legend()
plt.title("Denoising of Membrane Potential Signal Using PCA")
plt.show()
````
